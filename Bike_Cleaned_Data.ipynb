```python
#drop empty columns
df = df.drop(['Leg1-6', 'Leg2-6', 'Leg3-6', 'Leg4-6', 'Leg5-6', 'Leg6-1','Leg6-2', 'Leg6-3', 'Leg6-4', 'Leg6-5', 'leg6_enter', 'leg6_exit', 'leg6_total'], axis = 1)
```

```python
df = df.replace(0, np.nan)
```

```python
df.columns
```

```python
#check the empty variables by column
df[df['latitude'].isnull()]
```

```python
# drop all empty rows where latitude is null
df = df.dropna(subset=['latitude'])
```

```python
df.shape
```

```python
# latitude: remove negative values
#df['latitude'] = df['latitude'].abs()

# comparison: remove '-' and convert to numeric
df['comparison'] = (
    df['comparison']
    .str.replace('-', '', regex=False)
)

df['comparison'] = pd.to_numeric(df['comparison'], errors='coerce')
```

```python
df.head(5)
```

```python
df.shape
```

```python
desc_split = df['description'].str.split(',', expand=True)
desc_split = desc_split.apply(lambda col: col.str.strip())

df = pd.concat([df, desc_split], axis=1)
```

```python
df.shape
```

```python
df.head(2)
```

```python
#rename some columns for ease
df = df.rename(columns = {'Count_Date': 'count_date',
                          'Leg1-2': 'leg1_2', 
                          'Leg1-3': 'leg1-3',
                          'Leg1-4': 'leg1-4',
                          'Leg1-5': 'leg1-5',
                          'Leg2-1': 'leg2-1',
                          'Leg2-3': 'leg2-3', 
                          'Leg2-4': 'leg2-4', 
                          'Leg2-5': 'leg2-5', 
                          'Leg3-1': 'leg3-1', 
                          'Leg3-2': 'leg3-2', 
                          'Leg3-4': 'leg3-4', 
                          'Leg3-5': 'leg3-5',
                          'Leg4-1': 'leg4-1', 
                          'Leg4-2': 'leg4-2', 
                          'Leg4-3': 'leg4-3', 
                          'Leg4-5': 'leg4-5', 
                          'Leg5-1': 'leg5-1', 
                          'Leg5-2': 'leg5-2', 
                          'Leg5-3': 'leg5-3',
                          'Leg5-4': 'leg5-4', 
                          'Legs': 'legs', 
                          'State': 'state',
                          'Trips/hour': 'trips/hour',
                          'not known': 'not_known',
                           0: 'des_1',
                           1: 'des_2',
                           2: 'des_3',
                           3: 'des_4',
                           4: 'des_5'
                         })
```

```python
df.head(2)
```

```python
df = df.drop_duplicates(keep='first')
```

```python
#check for multiple values that are duplicated
df.query('count_date == "longitude"')
```

```python
df = df.fillna(0)
```

```python
df.head(5)
```

```python
df.dtypes
```

```python
#convert to the required data types
df['count_date'] = pd.to_datetime(df['count_date'], dayfirst=True, errors='coerce')
df['site_id'] = df['site_id'].astype(int)
df['not_known'] = df['not_known'].astype(int)
df['des_4'] = df['des_4'].astype(str)
df['des_5'] = df['des_5'].astype(str)
```
