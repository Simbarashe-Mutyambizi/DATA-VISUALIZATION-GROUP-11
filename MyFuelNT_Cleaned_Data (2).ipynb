```python
#rename the columns for ease
df = df.rename(columns = {'FullDate': 'full_date',
                         'Brand Name': 'brand_name',
                         'Region Name': 'region_name',
                         'Suburb': 'suburb',
                         'Postcode': 'postcode',
                         'Lat': 'lat',
                         'Long': 'long',
                         'Diesel': 'diesel',
                         'Premium 98': 'premium_98',
                         'Premium 95': 'premium_95',
                         'Unleaded 91': 'unleaded_91',
                         'Premium Diesel': 'premium_diesel',
                         'Low Aromatic Fuel': 'low_aromatic_fuel',
                         'Ethanol 105 (E85)': 'ethanol_105_E85',
                         'Ethanol 94 (E10)': 'ethanol_94_E10',
                         'Bio Diesel 20': 'bio_diesel_20'
                        })
```

```python
df.shape
```

```python
df['full_date'] = pd.to_datetime(df['full_date'], dayfirst=True, errors='coerce')
```

```python
bad = df[df['full_date'].isna()]
bad[['full_date']].head()
```

```python
df['day'] = df['full_date'].dt.day.astype('Int64')
df['month'] = df['full_date'].dt.month.astype('Int64')
df['year'] = df['full_date'].dt.year.astype('Int64')
```

```python
#convert to the required data types
df['brand_name'] = df['brand_name'].astype(str)
df['region_name'] = df['region_name'].astype(str)
df['suburb'] = df['suburb'].astype(str)
df['diesel'] = df['diesel'].astype(float)
df['premium_98'] = df['premium_98'].astype(float)
df['premium_95'] = df['premium_95'].astype(float)
df['unleaded_91'] = df['unleaded_91'].astype(float)
df['premium_diesel'] = df['premium_diesel'].astype(float)
df['LPG'] = df['LPG'].astype(float)
df['low_aromatic_fuel'] = df['low_aromatic_fuel'].astype(float)
df['ethanol_105_E85'] = df['ethanol_105_E85'].astype(float)
df['ethanol_94_E10'] = df['ethanol_94_E10'].astype(float)
```
